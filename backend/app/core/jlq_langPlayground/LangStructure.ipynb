```python
# qwen给出的野路子 ✅
from langchain_ollama import ChatOllama

# 初始化模型（需确保模型已通过 Ollama 下载）
llm = ChatOllama(
    model="gemma4:latest",          # 与 ollama list 中的名称一致
    base_url="http://localhost:11434",  # 默认地址可省略
    temperature=0.7            # 控制生成随机性
)

# 发送请求
response = llm.invoke("解释量子计算的基本原理")
print(response.content)
```

```python
# 尝试openai兼容的接口。。。✅
from langchain.agents import create_agent
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="gemma4:latest",
    base_url="http://localhost:11434/v1",   # Ollama 的 OpenAI 兼容端点
    api_key="ollama",                        # 随便填，Ollama 不验证
    temperature=0.7,
)

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model=llm,
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content)
```

```python
# 官网文档 不要用这个！❌
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

# 手动创建模型实例
llm = ChatOllama(
    model="gemma4:latest",
    temperature=0.7,
    streaming=False,          # 关闭流式，但没有真的关闭...
    # 底层发给ollama的http请求体中，stream字段仍然是true
    # 这是 langchain-ollama 的实现方式——它总是用流式请求，然后在 Python 端聚合结果。
    # 而 Gemma4 + Ollama 原生 POST/api/chat + stream: true + tools 这个组合，在 Ollama 服务端存在 bug，导致推理过程中崩溃返回 502。
)

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model=llm,  # 传入实例而非字符串
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)

result = agent.invoke(
    {"messages": [{"role": "user", "content": "What's the weather in San Francisco?"}]}
)
print(result["messages"][-1].content_blocks)
```
